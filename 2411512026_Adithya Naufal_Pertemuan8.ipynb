```python
import threading  # modul untuk menjalankan beberapa tugas sekaligus (multithreading)

lock = threading.RLock()  # RLock boleh dikunci berkali-kali oleh thread yang sama tanpa membuatnya macet

def rekursif(n): # fungsi rekursif
    with lock:  # ambil lock RLock sebelum lanjut
        print(f"masuk level{n}")  # tampilkan level rekursi saat ini
        if n>0:  # selama n masih lebih dari 0...
            rekursif(n-1)  # panggil fungsi ini lagi dengan n dikurangi 1

t = threading.Thread(target=rekursif, args=(2,))  # buat thread baru untuk menjalankan fungsi rekursif, mulai dari angka 2
t.start()  # jalankan thread-nya
t.join()  # tunggu sampai thread-nya benar-benar selesai
print("selesai tanpa deadlock (berkat RLock)")  # pesan ini muncul karena program tidak macet, berkat RLock
```

```python
import threading  # modul untuk menjalankan beberapa tugas (thread) sekaligus
import time       # modul untuk memberi jeda waktu pada program

sem = threading.Semaphore(1)  # Semaphore menentukan berapa banyak thread yang boleh masuk bersamaan; di sini nilainya 1

def gunakan_slot(nama):
    print(f"[{nama}] menunggu slot...")  # beri tahu bahwa thread ini sedang antre minta izin masuk
    with sem:  # minta izin ke Semaphore; kalau penuh, thread menunggu di sini sampai ada yang keluar
        print(f"[{nama}] MENDAPAT slot, bekerja selama 0.3 detik...")  # thread berhasil masuk dan mulai bekerja
        time.sleep(0.3)  # pura-pura bekerja selama 0.3 detik
        print(f"[{nama}] selesai, melepas slot")  # thread selesai dan mengembalikan slot ke Semaphore

threads = [threading.Thread(target=gunakan_slot, args=(f"T{i}",))for i in range(5)]  # buat 5 thread (T0 sampai T4) yang semuanya menjalankan fungsi gunakan_slot
for t in threads:
    t.start()  # jalankan semua thread sekaligus, jadi mereka berebut masuk ke Semaphore
for t in threads:
    t.join()   # tunggu sampai semua thread selesai

print("\nPerhatikan: meski ada 5 thread, HANYA 2 yang bekerja bersamaan pada satu waktu.")  # Catatan: karena Semaphore di atas diisi 1, sebenarnya hanya 1 thread yang bisa bekerja dalam satu waktu (bukan 2 seperti tertulis di pesan)
```

```python
import threading  # modul untuk menjalankan beberapa tugas (thread) sekaligus
import time       # modul untuk memberi jeda waktu

kondisi = threading.Condition()  # Condition dipakai agar thread bisa saling menunggu dan saling memberi kabar
status = {"siap": False}         # data bersama antar-thread; awalnya belum siap (False)

def penunggu(nama):
    with kondisi:                # ambil lock milik Condition sebelum mengecek atau menunggu
        print(f"[{nama}] menunggu status siap...")  # beri tahu bahwa thread ini mulai menunggu
        # wait_for() melepas lock sementara selama menunggu,
        # lalu mengambilnya lagi otomatis begitu kondisinya terpenuhi (True) dan ada notify
        kondisi.wait_for(lambda: status["siap"])  # thread tidur di sini sampai status["siap"] berubah jadi True
        print(f"[{nama}] status sudah siap, lanjut kerja!")  # thread terbangun dan melanjutkan kerjanya

def pemberi_sinyal():
    time.sleep(0.5)              # tunggu 0.5 detik sebelum mengirim sinyal
    with kondisi:                # ambil lock Condition untuk mengubah status bersama
        print("[Pemberi sinyal] mengubah status jadi siap dan notify_all()")  # info bahwa sinyal akan segera dikirim
        status["siap"] = True    # ubah status menjadi siap
        kondisi.notify_all()     # bangunkan SEMUA thread yang sedang menunggu supaya mereka mengecek kondisi lagi

# Buat 3 thread penunggu (Penunggu-0 sampai Penunggu-2)
threads = [threading.Thread(target=penunggu, args=(f"Penunggu-{i}",)) for i in range(3)]
# Buat 1 thread khusus sebagai pemberi sinyal
t_sinyal = threading.Thread(target=pemberi_sinyal)

for t in threads:
    t.start()                    # jalankan semua penunggu; mereka langsung antre menunggu
t_sinyal.start()                 # jalankan pemberi sinyal (menunggu 0.5 detik, lalu mengubah status)
for t in threads:
    t.join()                     # tunggu semua thread penunggu selesai
t_sinyal.join()                  # tunggu thread pemberi sinyal selesai
```

```python
import threading
import time
import random

lock_a = threading.Lock()
lock_b = threading.Lock()

def kerja_aman(nama, lock_pertama, lock_kedua, max_percobaan=10):
    for percobaan in range(1, max_percobaan + 1):
        if lock_pertama.acquire(timeout=0.2):
            try:
                time.sleep(0.05)  # pura-pura bekerja sebentar sambil memegang lock pertama
                
                if lock_kedua.acquire(timeout=0.2):
                    try:
                        print(f"[{nama}] berhasil dapat KEDUA lock pada percobaan ke-{percobaan}")
                        return
                    finally:
                        lock_kedua.release()  # pastikan lock kedua selalu dilepas kalau tadi sempat didapat
                else:
                    print(f"[{nama}] gagal dapat lock kedua (percobaan {percobaan}), lepas lock pertama & coba lagi")
            
            finally:
                lock_pertama.release()  # pastikan lock pertama selalu dilepas di akhir tiap percobaan
                
        time.sleep(random.uniform(0.01, 0.05))  # jeda acak sebelum mencoba lagi
        
    print(f"[{nama}] MENYERAH setelah {max_percobaan} percobaan")

# Thread 1 dan Thread 2 mengambil lock dengan urutan yang berlawanan, sehingga berpotensi deadlock; ini dicegah dengan timeout dan percobaan ulang
t1 = threading.Thread(target=kerja_aman, args=("Thread-1", lock_a, lock_b))
t2 = threading.Thread(target=kerja_aman, args=("Thread-2", lock_b, lock_a))

t1.start()
t2.start()

t1.join()
t2.join()

print("\nKeduanya selesai TANPA deadlock permanen, meski urutan pengambilan lock berlawanan!")
```

```python
import threading  # modul untuk mengelola beberapa thread
import time       # modul untuk memberi jeda waktu

lock_status = threading.Lock()  # lock untuk mengamankan pengecekan status lapar yang dipakai bersama
lapar = {"A": True, "B": True}  # status awal: A dan B sama-sama lapar (True)

def makan_sopan(nama, lawan, max_percobaan=8):  # simulasi thread yang "sopan" dan saling mengalah, dengan batas 8 kali percobaan
    percobaan = 0  # hitung sudah berapa kali mencoba
    while lapar[nama] and percobaan < max_percobaan:  # ulangi selama masih lapar dan belum melewati batas percobaan
        percobaan += 1  # tambah hitungan percobaan setiap putaran
        with lock_status:  # kunci akses supaya status lawan bisa dicek dengan aman
            if lapar[lawan]:  # kalau lawan masih lapar...
                print(f"[{nama}] percobaan {percobaan}: {lawan} masih lapar, saya mengalah dulu...")  # tampilkan pesan bahwa thread ini mengalah
            else:  # kalau lawan sudah selesai makan...
                print(f"[{nama}] percobaan {percobaan}: {lawan} sudah selesai, saya makan sekarang!")  # tampilkan pesan bahwa thread ini sekarang makan
                lapar[nama] = False  # ubah status diri sendiri jadi tidak lapar lagi
                continue  # lanjut ke putaran berikutnya
        time.sleep(0.1) # jeda sebentar; di sinilah kedua thread terus mengalah bersamaan sehingga terjadi livelock
    if lapar[nama]:  # kalau jatah percobaan habis dan masih lapar...
        print(f"[{nama}] MENYERAH setelah {max_percobaan} percobaan (LIVELOCK, tidak pernah makan!)")  # tampilkan pesan menyerah karena livelock (tidak pernah makan)

# Buat dua thread (t1 untuk A, t2 untuk B) yang saling mengalah
t1 = threading.Thread(target=makan_sopan, args=("A", "B"))
t2 = threading.Thread(target=makan_sopan, args=("B", "A"))
t1.start(); t2.start()  # jalankan kedua thread bersamaan
t1.join(); t2.join()    # tunggu sampai kedua thread selesai atau menyerah
```

```python
import threading  # modul untuk mengelola beberapa thread
import time       # modul untuk memberi jeda waktu
import random     # modul angka acak, dipakai untuk memecah kondisi saling mengalah yang simetris

lock_status = threading.Lock()  # lock agar pengecekan status dilakukan bergantian
lapar = {"A": True, "B": True}  # status awal: A dan B sama-sama lapar

def makan_sopan_v2(nama, lawan, max_percobaan=8):  # versi perbaikan livelock, dengan batas 8 kali percobaan
    percobaan = 0  # hitung sudah berapa kali mencoba
    while lapar[nama] and percobaan < max_percobaan:  # ulangi selama masih lapar dan belum melewati batas percobaan
        percobaan += 1  # tambah hitungan percobaan
        with lock_status:  # kunci akses untuk membaca/mengubah status bersama
            # Pemecah kebuntuan acak: kalau lawan masih lapar, 70% kemungkinan mengalah dan 30% kemungkinan nekat makan
            if lapar[lawan] and random.random() > 0.3:
                print(f"[{nama}] percobaan {percobaan}: {lawan} masih lapar, saya mengalah dulu...")
            else:  # kalau kebagian giliran nekat, atau lawan sudah selesai makan...
                print(f"[{nama}] percobaan {percobaan}: saya makan sekarang!")
                lapar[nama] = False  # ubah status diri sendiri jadi tidak lapar lagi
                continue  # lanjut ke putaran berikutnya
        time.sleep(0.05)  # jeda sebentar sebelum mencoba lagi
    if lapar[nama]:  # kalau jatah percobaan habis dan masih lapar...
        print(f"[{nama}] MENYERAH setelah {max_percobaan} percobaan")

random.seed(3)  # kunci pola angka acak supaya hasilnya sama setiap kali dijalankan
lapar["A"], lapar["B"] = True, True  # reset status A dan B menjadi lapar lagi

# Buat dua thread untuk A dan B memakai fungsi versi 2
t1 = threading.Thread(target=makan_sopan_v2, args=("A", "B"))
t2 = threading.Thread(target=makan_sopan_v2, args=("B", "A"))

t1.start(); t2.start()  # jalankan kedua thread bersamaan
t1.join(); t2.join()    # tunggu sampai kedua thread selesai
print("\nDengan pemutus simetri, kebuntuan berhasil dipecahkan.")  # pesan sukses
```

```python
import threading  # modul untuk mengelola beberapa thread
import time       # modul untuk memberi jeda waktu

lock = threading.Lock()  # lock bersama yang diperebutkan kedua thread
hitungan = {"Rakus": 0, "Sabar": 0}  # hitung berapa kali masing-masing thread berhasil mendapat lock
berhenti = threading.Event()  # sinyal untuk menyuruh kedua thread berhenti nanti

def thread_rakus():
    while not berhenti.is_set():  # terus berjalan selama sinyal berhenti belum aktif
        with lock:  # ambil lock
            hitungan["Rakus"] += 1  # tambah jumlah perolehan lock
            # TIDAK ada jeda, jadi langsung mencoba lagi dan hampir selalu menang rebutan lock

def thread_sabar():
    while not berhenti.is_set():  # terus berjalan selama belum disuruh berhenti
        time.sleep(0.01)  # tidur 0.01 detik dulu sebelum mencoba lagi, supaya lebih "sopan"
        with lock:  # ambil lock
            hitungan["Sabar"] += 1  # tambah jumlah perolehan lock

t1 = threading.Thread(target=thread_rakus)  # buat thread Rakus
t2 = threading.Thread(target=thread_sabar)  # buat thread Sabar
t1.start(); t2.start()  # jalankan kedua thread bersamaan
time.sleep(1.0)  # biarkan keduanya berebut lock selama 1 detik penuh
berhenti.set()  # kirim sinyal agar kedua thread berhenti
t1.join(); t2.join()  # tunggu sampai kedua thread benar-benar berhenti

# Tampilkan statistik perolehan lock
print(f"Jumlah lock didapat 'Rakus' : {hitungan['Rakus']:,}")
print(f"Jumlah lock didapat 'Sabar' : {hitungan['Sabar']:,}")
print(f"\nRasio ketimpangan: 'Rakus' mendapat lock {hitungan['Rakus']/max(hitungan['Sabar'],1):.0f}x lebih sering")
print("Ini adalah STARVATION -- 'Sabar' secara teknis TIDAK deadlock (masih dapat giliran sesekali),")
print("tapi porsinya sangat tidak adil dibanding 'Rakus'.")
```

```python
import threading  # modul untuk mengelola beberapa thread
import time       # modul untuk memberi jeda waktu
import random     # modul untuk membuat variasi waktu secara acak

UKURAN_BUFFER = 3  # kapasitas maksimal buffer: 3 item
buffer = []        # wadah penyimpanan item (list kosong)

mutex = threading.Lock()                      # lock supaya akses ke buffer tidak bertabrakan
slot_kosong = threading.Semaphore(UKURAN_BUFFER)  # Semaphore penghitung slot kosong (awalnya 3)
slot_terisi = threading.Semaphore(0)              # Semaphore penghitung item yang siap diambil (awalnya 0)

def producer(jumlah_item):
    for i in range(jumlah_item):
        item = f"item-{i}"  # buat nama item baru
        time.sleep(random.uniform(0.02, 0.08))  # pura-pura sedang membuat item
        slot_kosong.acquire()  # tunggu sampai ada slot kosong; kalau buffer penuh, producer berhenti dulu
        with mutex:            # kunci buffer supaya aman saat diubah
            buffer.append(item)  # masukkan item baru ke buffer
            print(f"[Producer] taruh {item:<8} (buffer: {len(buffer)}/{UKURAN_BUFFER})")  # tampilkan status buffer
        slot_terisi.release()  # kabari consumer bahwa ada 1 item baru yang siap diambil

def consumer(jumlah_item, nama):
    for _ in range(jumlah_item):
        slot_terisi.acquire()  # tunggu sampai ada item di buffer; kalau kosong, consumer menunggu
        with mutex:            # kunci buffer saat mengambil item
            item = buffer.pop(0)  # ambil item paling depan dari buffer
            print(f"    [{nama}] ambil {item:<8} (buffer: {len(buffer)}/{UKURAN_BUFFER})")  # tampilkan status buffer
        slot_kosong.release()  # kabari producer bahwa ada 1 slot kosong baru

JUMLAH_ITEM = 9  # total item yang akan dibuat
t_p = threading.Thread(target=producer, args=(JUMLAH_ITEM,))  # buat thread producer
t_c1 = threading.Thread(target=consumer, args=(JUMLAH_ITEM // 2 + JUMLAH_ITEM % 2, "Consumer-1"))  # consumer pertama (mengambil 5 item)
t_c2 = threading.Thread(target=consumer, args=(JUMLAH_ITEM // 2, "Consumer-2"))                  # consumer kedua (mengambil 4 item)

t_p.start(); t_c1.start(); t_c2.start()  # jalankan semua thread bersamaan
t_p.join(); t_c1.join(); t_c2.join()    # tunggu sampai semua proses produksi dan konsumsi selesai

print(f"\nSisa buffer: {buffer} (harus kosong jika semua item terkonsumsi)")  # pengecekan akhir
```

```python
import threading  # modul untuk mengelola thread

class RekeningBank:
    def __init__(self, nama, saldo_awal):
        self.nama = nama           # simpan nama pemilik rekening
        self.saldo = saldo_awal    # simpan saldo awal rekening
        self.lock = threading.RLock()  # pakai RLock (reentrant lock) supaya thread yang sama boleh masuk ke method yang saling memanggil

    def setor(self, jumlah):
        with self.lock:            # kunci rekening sebelum mengubah saldo
            self.saldo += jumlah   # tambahkan ke saldo
            print(f"[{self.nama}] Setor {jumlah}, Saldo: {self.saldo}")

    def tarik(self, jumlah):
        with self.lock:            # kunci rekening sebelum mengubah saldo
            if self.saldo >= jumlah:       # cek apakah saldonya cukup
                self.saldo -= jumlah       # kurangi saldo
                print(f"[{self.nama}] Tarik {jumlah}, Saldo: {self.saldo}")
                return True                # kembalikan True kalau penarikan berhasil
            else:
                print(f"[{self.nama}] Tarik {jumlah} GAGAL (Saldo tidak cukup)")
                return False               # kembalikan False kalau penarikan gagal

    def transfer(self, rekening_tujuan, jumlah):
        with self.lock:            # kunci rekening asal (diri sendiri) selama proses transfer
            print(f"[{self.nama}] Memulai transfer {jumlah} ke {rekening_tujuan.nama}...")
            # Di sinilah RLock berguna! Method tarik() akan mengunci self.lock LAGI.
            # RLock mengizinkannya, sedangkan Lock biasa akan membuat program macet di sini (self-deadlock).
            if self.tarik(jumlah): 
                rekening_tujuan.setor(jumlah)  # setorkan uangnya ke rekening tujuan
                print(f"[{self.nama}] Transfer berhasil ke {rekening_tujuan.nama}")

# ---- PENGUJIAN ----
rek_A = RekeningBank("Andi", 1000)
rek_B = RekeningBank("Budi", 500)

# Jalankan transfer di dalam sebuah thread
t = threading.Thread(target=rek_A.transfer, args=(rek_B, 300))
t.start()
t.join()
```

```python
import subprocess
import sys

# Skrip ini berisi dua skenario: versi deadlock dan versi perbaikannya
skrip_tiga_lock = """\n""".join([
    "import threading, time",
    "lock_a = threading.Lock()",
    "lock_b = threading.Lock()",
    "lock_c = threading.Lock()",
    "",
    "def thread1():",
    "    with lock_a:",
    "        time.sleep(0.1)",
    "        with lock_b: print('T1 selesai')",
    "",
    "def thread2():",
    "    with lock_b:",
    "        time.sleep(0.1)",
    "        with lock_c: print('T2 selesai')",
    "",
    "def thread3_deadlock():",
    "    with lock_c:",                   # Ambil C lalu menunggu A -- ini yang memicu circular wait (saling menunggu melingkar)
    "        time.sleep(0.1)",
    "        with lock_a: print('T3 selesai')",
    "",
    "def thread3_fix():",
    "    with lock_a:",                   # PERBAIKAN: selalu ambil lock sesuai urutan abjad (A dulu, baru C)
    "        time.sleep(0.1)",
    "        with lock_c: print('T3 selesai')",
    "",
    "import sys",
    "mode = sys.argv[1] if len(sys.argv) > 1 else 'deadlock'",
    "target_t3 = thread3_deadlock if mode == 'deadlock' else thread3_fix",
    "",
    "t1 = threading.Thread(target=thread1)",
    "t2 = threading.Thread(target=thread2)",
    "t3 = threading.Thread(target=target_t3)",
    "t1.start(); t2.start(); t3.start()",
    "t1.join(); t2.join(); t3.join()",
    "print(f'STATUS: {mode.upper()} berhasil dilewati!')"
])

# Simpan skrip ke file
with open("demo_3lock.py", "w") as f:
    f.write(skrip_tiga_lock)

# 1. Uji versi deadlock
print("--- MENGUJI VERSI DEADLOCK ---")
try:
    hasil_macet = subprocess.run([sys.executable, "demo_3lock.py", "deadlock"], capture_output=True, text=True, timeout=2)
except subprocess.TimeoutExpired:
    print("TIMEOUT 2 DETIK: Terjadi Deadlock 3 Arah!")

# 2. Uji versi yang sudah diperbaiki
print("\n--- MENGUJI VERSI DIPERBAIKI ---")
hasil_lancar = subprocess.run([sys.executable, "demo_3lock.py", "fix"], capture_output=True, text=True, timeout=2)
print(hasil_lancar.stdout.strip())
```

```python
import threading
import time
import random

def uji_buffer(ukuran):
    buffer = []
    mutex = threading.Lock()
    slot_kosong = threading.Semaphore(ukuran)   # jumlah slot Semaphore disesuaikan dengan ukuran buffer (1, 3, atau 10)
    slot_terisi = threading.Semaphore(0)

    def producer(id_p):
        for i in range(3):                      # setiap producer membuat 3 barang
            item = f"P{id_p}-{i}"
            time.sleep(random.uniform(0.01, 0.05))
            slot_kosong.acquire()               # minta slot kosong
            with mutex:
                buffer.append(item)
                print(f"[Prod-{id_p}] Taruh {item} (Isi: {len(buffer)}/{ukuran})")
            slot_terisi.release()               # kabari bahwa ada barang masuk

    def consumer(id_c):
        for _ in range(3):                      # setiap consumer mengambil 3 barang
            slot_terisi.acquire()               # minta barang
            with mutex:
                item = buffer.pop(0)
                print(f"   [Cons-{id_c}] Ambil {item} (Isi: {len(buffer)}/{ukuran})")
            slot_kosong.release()               # kabari bahwa ada slot kosong

    # Buat 2 producer dan 2 consumer
    threads = [
        threading.Thread(target=producer, args=(1,)),
        threading.Thread(target=producer, args=(2,)),
        threading.Thread(target=consumer, args=(1,)),
        threading.Thread(target=consumer, args=(2,))
    ]
    
    print(f"\n=== MENGUJI DENGAN UKURAN_BUFFER = {ukuran} ===")
    for t in threads: t.start()
    for t in threads: t.join()

uji_buffer(1)
uji_buffer(3)
uji_buffer(10)
```
